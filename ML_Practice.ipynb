```python
import pandas as pd

melbourne_file_path = '/content/drive/MyDrive/Data Science/Datasets/melb_data.csv'
df = pd.read_csv(melbourne_file_path)
df.columns
# df = df.dropna(axis=0)

```

```python
y = df.Price
features = ['Rooms','Bathroom','Landsize','BuildingArea',
                        'YearBuilt','Lattitude','Longtitude']
X = df[features]

X.describe()
```

```python
X.head()
```

```python
from sklearn.tree import DecisionTreeRegressor

melbourne_model = DecisionTreeRegressor(random_state=1)

melbourne_model.fit(X,y)
```

```python
print("Making predictions for the following 5 houses:")
print(X)
print("The predictions are:")
print(melbourne_model.predict(X))
```

# Model Validation #


```python
# How to calculate mean_absolute_error
from sklearn.metrics import mean_absolute_error
predicted_home_prices = melbourne_model.predict(X)
mean_absolute_error(y,predicted_home_prices)   # in-sample - not recommended, very inaccurate when used in practice
```

```python
from sklearn.model_selection import train_test_split
# split data into training and validation data, for both features and target
train_X, val_X, train_y, val_y = train_test_split(X,y,random_state=0)

melbourne_model = DecisionTreeRegressor()
melbourne_model.fit(train_X,train_y)

val_predictions = melbourne_model.predict(val_X)
print(mean_absolute_error(val_y,val_predictions))
```

# Random Forest #


```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

forest_model = RandomForestRegressor(random_state=1)
forest_model.fit(train_X,train_y)
melb_preds = forest_model.predict(val_X)
print(mean_absolute_error(val_y, melb_preds))
```
